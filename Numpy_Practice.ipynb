```python
#In order to measure speed and to find which onw is fast we will import time module
import time
```

```python
#we are making two lists,containing 1lac elements in each.And we try to map each element of first
#list with the first element of second list and so on.
x=range(100000)
y=range(100000,200000)
#first time is a module or library and second time is function of that module and library.It tells us
#about the current time,we will store that current time in start-time variable
start_time=time.time()
#we will make list by adding them and storing in a seperate variable
c=[(x,y) for x,y in zip(x,y)] #zip make zip of both by combining them
#to measue speed we will subtract the time now and the start time we save in start-time variable
print(time.time()-start_time)
```

```python
#now performing same on numpy array
a=np.arange(100000)
b=np.arange(100000,200000)
start_time=time.time()
c=a+b #combining both arrays
print(time.time()-start_time)
#not a big difference BUTTTTT IF WE CHANGEEEE THE VALUE LITTLE BIT
```
